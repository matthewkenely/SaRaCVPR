```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

import os
import tqdm
```

```python
# Create new directory called ASSRBinary with the same structure as ASSR
# ASSRBinary will contain the binary images of the ASSR images

ASSR_PATH = './ASSR/'
ASSR_BINARY_PATH = './ASSRBinary/'

if not os.path.exists(ASSR_BINARY_PATH):
    os.makedirs(ASSR_BINARY_PATH)

# ASSRBinary should contain folders gt and images. Each of those should contain test, train and val. Create directories
dirs = ['gt', 'images']
subdirs = ['test', 'train', 'val']

for d in dirs:
    for sd in subdirs:
        path = ASSR_BINARY_PATH + d + '/' + sd + '/'
        if not os.path.exists(path):
            os.makedirs(path)

# Load the images from the ASSR dataset in the same locations
# For each image, convert it to binary. Use tqdm to show progress

kernel = np.ones((11, 11),np.uint8)
for d in dirs:
    for sd in subdirs:
        path = ASSR_PATH + d + '/' + sd + '/'
        files = os.listdir(path)

        for f in tqdm.tqdm(files):
            img = cv2.imread(path + f, cv2.IMREAD_GRAYSCALE)

            # erode to add border between objects
            img = cv2.erode(img, kernel, iterations=1)
                
            img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)[1]
            cv2.imwrite(ASSR_BINARY_PATH + d + '/' + sd + '/' + f, img)
            

    
```
